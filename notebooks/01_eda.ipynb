## ClosedEvent & CLOSED_BY Relationship

Issue có thể có `ClosedEvent`, nhưng `closer` không phải lúc nào cũng được API cung cấp.

Có 2 trường hợp:

### 1. `closer != null`

API cung cấp thông tin về object đã close Issue:

~~~python
{
    "__typename": "ClosedEvent",
    "closer": {
        "number": 45,
        "title": "feat: rename Vault to OpenBao in build files",
        "url": "https://github.com/openbao/openbao/pull/45"
    }
}
~~~

→ Có thể xác định `CLOSED_BY` relationship.

### 2. `closer == null`

~~~python
{
    "__typename": "ClosedEvent",
    "closer": None
}
~~~

→ Issue có `ClosedEvent`, nhưng API không cung cấp thông tin về `closer`.

**Không nên kết luận rằng Issue chắc chắn không được PR/commit đóng.**

---

## Database Design

Không được thiết kế:

~~~sql
closing_pr_id INTEGER NOT NULL
~~~

vì có những record không có `closer`.

Thay vào đó, `CLOSED_BY` là một **optional relationship**:

~~~text
Issue
  │
  └── CLOSED_BY ──→ Pull Request / Commit
        │
        └── có thể tồn tại hoặc không
~~~

Ví dụ:

~~~text
Issue #1 ── CLOSED_BY ──→ PR #45
Issue #2 ── CLOSED_BY ──→ NULL
Issue #3 ── CLOSED_BY ──→ PR #72
~~~

Nếu dùng relational database:

~~~sql
closing_pr_id INTEGER NULL
~~~

### Key takeaway

> `closer = null` means the API response does not provide the closer,
> not necessarily that no PR/commit closed the Issue.

`````python
# Inspect Closer thuộc loại gì 
for issue in data:
    for event in issue["timelineItems"]["nodes"]:
        if event["__typename"] == "ClosedEvent":
            closer = event["closer"]

            if closer:
                print(closer)
`````

# CrossReferencedEvent - Insights

## 1. CrossReferencedEvent cho biết một object khác đang reference Issue hiện tại

Cấu trúc:

~~~python
{
    "__typename": "CrossReferencedEvent",
    "willCloseTarget": False,
    "source": {
        "number": 2,
        "title": "Update go mod ref",
        "url": "https://github.com/openbao/openbao/pull/2"
    }
}
~~~

Có thể hiểu:

~~~text
Current Issue
     ↑
     │ referenced by
     │
source (#2)
~~~

=> `source` là object đã reference Issue hiện tại.

---

## 2. `source` có thể là Pull Request hoặc Issue

Trong data có các URL dạng:

~~~text
/pull/2
/pull/5
/pull/22
~~~

=> `source` là Pull Request.

Nhưng cũng có:

~~~text
/issues/1974
/issues/13
/issues/17
~~~

=> `source` là Issue.

Vì vậy:

~~~text
Issue
  ↑
  │ referenced by
  │
  ├── Pull Request
  │
  └── Issue
~~~

Không nên assume `source` luôn là Pull Request.

---

## 3. `CrossReferencedEvent` không đồng nghĩa với `CLOSED_BY`

Ví dụ:

~~~python
{
    "__typename": "CrossReferencedEvent",
    "willCloseTarget": False,
    "source": {...}
}
~~~

Ở đây:

~~~text
willCloseTarget = False
~~~

=> Source chỉ reference Issue, không được đánh dấu là sẽ close Issue.

Do đó:

~~~text
CrossReferencedEvent
        ≠
CLOSED_BY relationship
~~~

Nên phân biệt:

~~~text
PR ── REFERENCES ──→ Issue

PR ── CLOSED_BY ───→ Issue
~~~

Hai relationship này có ý nghĩa khác nhau.

---

## 4. Một source có thể xuất hiện nhiều lần

Trong data có trường hợp:

~~~text
PR #5
PR #5
~~~

xuất hiện nhiều `CrossReferencedEvent`.

Điều này cho thấy:

> Không nên assume một cặp source-target chỉ xuất hiện một lần trong timeline.

Có thể có nhiều event cho cùng một relationship:

~~~text
PR #5 ── REFERENCES ──→ Issue #X
PR #5 ── REFERENCES ──→ Issue #X
~~~

Do đó cần phân biệt:

~~~text
Timeline Event
      vs
Underlying Relationship
~~~

---

## 5. Cross-reference có thể đến từ repository khác

Trong data có:

~~~text
https://github.com/awkto/openbao-fork/issues/3
~~~

Trong khi phần lớn source đến từ:

~~~text
https://github.com/openbao/openbao/...
~~~

=> Cross-reference có thể xảy ra giữa các repository khác nhau.

Ví dụ:

~~~text
Repository A
    │
    │ REFERENCES
    ▼
Repository B
    │
    ▼
Issue
~~~

Vì vậy không thể chỉ dùng `issue_number` để identify một Issue.

Ví dụ:

~~~text
openbao/openbao#13
awkto/openbao-fork#13
~~~

có cùng `number = 13` nhưng là hai Issue khác nhau.

=> Identity nên bao gồm ít nhất:

~~~text
repository + issue_number
~~~

hoặc tốt hơn là dùng GitHub's global node ID nếu available.

# Database / Data Model Insight

Từ `ClosedEvent` và `CrossReferencedEvent`, không nên model dữ liệu chỉ đơn giản như:

~~~text
Issue
 └── closing_pr_id
~~~

Thay vào đó nên nghĩ theo hướng relationship / graph:

~~~text
Issue
  │
  ├── CLOSED_BY ──────→ Pull Request / Commit
  │
  └── REFERENCED_BY ──→ Pull Request / Issue
                              │
                              └── có thể ở repository khác
~~~

Timeline event layer:

~~~text
TimelineEvent
│
├── ClosedEvent
│     └── closer: Pull Request | Commit | null
│
└── CrossReferencedEvent
      ├── source: Pull Request | Issue
      └── willCloseTarget: boolean
~~~

## Key Takeaways

1. `CrossReferencedEvent` = có object khác reference Issue.
2. `source` có thể là PR hoặc Issue.
3. `CrossReferencedEvent` không đồng nghĩa với `CLOSED_BY`.
4. `willCloseTarget` là một signal riêng cần giữ lại.
5. Một source có thể xuất hiện nhiều lần trong timeline.
6. Cross-reference có thể xảy ra giữa các repository khác nhau.
7. Không dùng `issue_number` đơn độc làm identity.
8. Nên model các relationship như `REFERENCES` và `CLOSED_BY` riêng biệt.
9. `ClosedEvent.closer` có thể là PR, Commit hoặc `null`.
10. Timeline Event và underlying Relationship nên được phân biệt khi thiết kế database.
